```python
#5 .Pallindrome

lst = [1,2,3,2,1]

rev = []
for i in lst:
    rev = [i]+rev
print(rev)

if lst == rev:
    print("pallindrome")
else:
    print("Not Pallindrome")    
```

```python
#11. Sort a List Without using the built-in sort function
lst = [8,2,4,9,3,1]
sort_lst = []

while lst:                       #1st Method
    minimum = min(lst)
    sort_lst.append(minimum)
    lst.remove(minimum)
print(sort_lst)
```

```python
lst = [8, 2, 4, 9, 3, 1]
sort_lst = []                      #2nd Method

for num in range(len(lst)):
    maximum = lst[0]
    
    for i in lst:
        if i < maximum:
            maximum = i

    sort_lst.append(maximum)
    lst.remove(maximum)

print(sort_lst)
```

```python
#13. Count Even and Odd in a List

lst = [1,2,3,4,5,6,7,8]
even = []
odd = []

for num in lst:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)
print("Even Numbers = ",len(even))
print("Odd Numbers =",len(odd))
```

```python
#17. Find a Missing number from a list of 1 to n

lst = [1,2,4,5,6,8,9,10]
n = len(lst)+1
expected_sum = n*(n+1)//2
actual_sum = 0
for i in lst:
    actual_sum+=i
missing_num = expected_sum-actual_sum
print(missing_num)
    
```

```python
#19. Find the Intersection of Two List

a = [1,2,3,4,5]
b = [3,4,5,6,7]
intersec = []

for i in a:
    if i in b:
        intersec.append(i)


print(intersec)
```

```python
#20. Find the Union of two list

a = [1,2,3,4,5]
b = [3,4,5,6,7]

uni = []
for i in a:
    if i not in uni:
        uni.append(i)

for i in b:
    if i not in uni:
        uni.append(i)
print(uni)
```
