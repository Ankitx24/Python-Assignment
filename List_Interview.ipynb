```python
#5 .Pallindrome

lst = [1,2,3,2,1]

rev = []
for i in lst:
    rev = [i]+rev
print(rev)

if lst == rev:
    print("pallindrome")
else:
    print("Not Pallindrome")    
```

```python
#11. Sort a List Without using the built-in sort function
lst = [8,2,4,9,3,1]
sort_lst = []

while lst:                       #1st Method
    minimum = min(lst)
    sort_lst.append(minimum)
    lst.remove(minimum)
print(sort_lst)
```

```python
lst = [8, 2, 4, 9, 3, 1]
sort_lst = []                      #2nd Method

for num in range(len(lst)):
    maximum = lst[0]
    
    for i in lst:
        if i < maximum:
            maximum = i

    sort_lst.append(maximum)
    lst.remove(maximum)

print(sort_lst)
```

```python
#20. Find the Union of two list

a = [1,2,3,4,5]
b = [3,4,5,6,7]

uni = []
for i in a:
    if i not in uni:
        uni.append(i)

for i in b:
    if i not in uni:
        uni.append(i)
print(uni)
```

```python
#23. Find a first non -repeating element in a list

arr = [1,2,3,3,4,5,3,4,7,8,1]
# for i in arr:
#     if arr.count(i) == 1:
#         print(i)

freq = {}
for i in arr:
    if i in freq:
        freq[i]+=1
    else:
        freq[i] = 1
print(freq)

for key,value in freq.items():
    if value == 1:
        print(key)
        break
```

```python
#24. find the majority value in a list.
arr = [1,2,3,3,4,5,3,4,7,8,1]

freq = {}

for i in arr:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

print(freq)

max_freq = max(freq.values())

for key, value in freq.items():
    if value == max_freq:
        print("Element with maximum frequency:", key)
        break
```
