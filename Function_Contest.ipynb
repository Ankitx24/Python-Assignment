Q.1 A leading e-commerce company, "QuickBuy", has a vast customer base and offers various products to its customers. The company wants to analyze the purchasing behavior of its customers based on their demographics and product categories. They have collected a dataset containing customer information such as age, gender, occupation, and location along with their purchase history. The dataset has around 10,000 rows and 15 columns.

The data analysis team at QuickBuy wants to develop a Python function that can help them calculate the average purchase amount spent by customers in each age group (18-25, 26-35, 36-45, 46-55, and above 55) on specific product categories (Electronics, Fashion, Home Appliances).

Write a Python function named calculate_average_purchase that takes two parameters -customer data (a pandas DataFrame containing customer data) and product category (a string specifying the product category). The function should return a dictionary where the keys are the age groups and the values are the average purchase amounts for each age group on the specified product category.

For example, if we call calculate_average_purchase (customer_data, 'Electronics'), it should return a dictionary like (18-25: 5000.0, 26-35: 6000.0, ...}.

You can assume that the customer data is stored in a pandas DataFrame with columns -'Age', 'Gender', 'Occupation', 'Location', 'Product_Category', and 'Purchase_Amount'.

import pandas as pd

#assume customer data is a pandas DataFrame.

```python
def calculate_average_purchase(customer_data,product_category):
    df = customer_data[customer_data['Product_Category'] == product_category]

    bins = [18,25,35,45,55,float('inf')]
    labels = ['18-25','26-35','36-45','46-55','55+']

    df['Age_Group'] = pd.cut(df['Age'], bins = bins, labels = labels, right = True)

    result = df.groupby('Age_Group'['Purchase_Amount'].mean())


    return result.to_dict()
```

Q3.Write a function that counts the number of vowels in a word. For example:

"apple" -> 2 vowels (a, e)

"strength"-> 1 vowel (c)

"QUEUE"-> 4 vowels (U, E, U, E)

"xyz"-> 0 vowels

The function should be case-insensitive (i.e., "A" and "a" are both counted as vowels)

```python
def count_vowels(word):
    vowels = "aeiou"
    count = 0

    for ch in word.lower():
        if ch in vowels:
            count += 1

    return count        
```
