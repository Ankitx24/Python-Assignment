# concatenatee and merge dataframe(JOINS)

```python
df1 = pd.DataFrame({'ID':[1,2,3],'Name':['A','B','C']})
df1
```

```python
df2 = pd.DataFrame({'ID':[1,2,3,4],'Score':[88,99,77,79]})
df2
```

```python
# Vertical / Row Level / top on top
pd.concat([df1,df2],axis = 0)  
```

```python
# Horizontal/Column level / Side on Side
pd.concat([df1,df2],axis=1)
```

```python
# merge = join (inner,left,right,outer)
pd.merge(df1,df2,how='inner',on='ID')
```

```python

pd.merge(df1,df2,how='inner',left_on='ID',right_on='ID')
```
