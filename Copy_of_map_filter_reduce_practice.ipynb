# Map, Filter, Reduce

## Problem 1: Square all numbers using `map()`

**Approach:**
- Take a list of numbers
- Use `map()` to apply square operation on each element
- Convert result to list

```python
l = [1,2,3,4,5]

sq = list(map(lambda x:x**2, l))

Sq_list = list(sq)

print(Sq_list)
```

## Problem 2: Convert list of strings to uppercase

**Approach:**
- Use `map()`
- Apply `upper()` on each string

```python
n = ["apple","orange","banana","pineapple"]

n_upper = list(map(lambda x:x.upper(), n))

print(n_upper)
```

## Problem 3: Filter even numbers using `filter()`

**Approach:**
- Use `filter()`
- Condition: number % 2 == 0

```python
number = [1,2,3,4,5,6,7,8]

even_num = list(filter(lambda x:x%2==0, number))

print(even_num)
```

## Problem 4: Filter names longer than 4 characters

**Approach:**
- Use `filter()`
- Keep strings with length > 4

```python
names = ["Kalia" ,"Avi" ,"Bheem" ,"Raju"]

long_names = list(filter(lambda x:len(x) > 4, names))

print(long_names)
```

## Problem 5: Find sum of numbers using `reduce()`

**Approach:**
- Import reduce from functools
- Add elements step-by-step

```python
from functools import reduce
def add(x,y):
  return x+y

a = [1,2,3,4,5]
sum = reduce(add,a)
print(sum)
```

## Problem 6: Find product of numbers

**Approach:**
- Use reduce
- Multiply elements

```python
from functools import reduce
def mul(x,y):
  return x*y

a = [1,2,3,4,5]
multiply = reduce(mul,a)
print(multiply)
```

## Problem 7: Combined Map + Filter

**Problem:** Square numbers and keep only even results

**Approach:**
- Map to square
- Filter even values

```python
l = [1,2,3,4,5,6,7,8]

Sq_even = list(filter(lambda x:x%2==0 ,map(lambda x:x**2, l)))

print(Sq_even)
```
